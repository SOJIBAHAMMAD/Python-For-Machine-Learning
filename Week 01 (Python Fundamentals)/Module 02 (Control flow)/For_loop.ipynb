```python
# for loop variation-1
for i in range(3):
    print(i,"Hello World!")
```

```python
# for loop variation-2
for i in range(1,5):
    print(i,"SAJIB AHAMMAD")
```

```python
# for loop variation-3
for i in range(1,5,1):
    print(i,"SAJIB AHAMMAD")
```

```python
# nested for loop variation-4
for i in range(1,5,1):
    for j in range(1,4,1):
        print(i,j)
```
