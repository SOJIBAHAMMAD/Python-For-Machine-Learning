```python
inp = input()
numbers = inp.split()

x = int(numbers[0])
y = int(numbers[1])
z = int(numbers[2])

if (x>y and x>z):
    if(y>z):
        print(z,x)
    else:
        print(y,x)
elif (y>x and y>z):
    if(x>z):
        print(z,y)
    else:
        print(x,y)
else:
    if(x>y):
        print(y,z)
    else:
        print(x,z)

```
