```python
## List Dicleration
numbers = [10,20,30,40,50] ## int
float_numbers = [10.31,33.12,40.42,20.95]  ## Float
fruits = ['Apple','Banana','Lichi','Mango']  ## String
mix_type = [10,20.32,'Apple']  ## Mix type
list_er_modde_list = [ [10,20,30],[10.21,20.32,30.33],['Apple','Banana','Lichi']]

print(numbers)
print(float_numbers)
print(fruits)
print(mix_type)
print(list_er_modde_list)
print(list_er_modde_list[2][1])
## Access
print(numbers[2])
# List is mutable cz but String is not
numbers[2] = 60
print(numbers[2])
## New list
new_list = numbers[0:3]
print(new_list)
```

## Modification

```python
## Adding an element
numbers.append(70)
print(numbers)

## Insertion at any position
numbers.insert(3,80)
print(numbers)

## Deleting the last value
numbers.pop()
print(numbers)

## Deleting a specific value
numbers.remove(60)
print(numbers)
```

```python
## List Sort Assending
numbers.sort()
print(numbers)

## List Sort Desending
numbers.reverse()
print(numbers)
##Or
numbers.sort(reverse=True)
print(numbers)
```

## List as Stack

```python
stack = []
# Pushing in Stack
stack.append(1)
stack.append(2)
stack.append(3)
stack.append(4)
stack.append(5)
print(stack)

## Getting the top element
print(f"The top element is {stack[-1]}")
## Removing the top element
print(stack.pop())
print(f"The top element is {stack[-1]}")
print(stack.pop())
```

## Stack as Queue

```python
queue = []
## Pushing in Queue
queue.append(1)
queue.append(2)
queue.append(3)
queue.append(4)
queue.append(5)

## Getting the top element
print(f"The top element is {queue[-1]}")
## Removing front element
print(queue.pop(0))
print(f"Now front element is {queue[0]}")
```
