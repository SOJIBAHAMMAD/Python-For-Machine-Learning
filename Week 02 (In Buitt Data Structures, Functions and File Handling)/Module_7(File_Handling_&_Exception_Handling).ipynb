Practice 1 :
Get the text from sample.txt , and count the following

Number of lines

Number of words

Number of characters

and save it to the counter_of_string.txt file

```python
from functools import reduce
with open("./sample_data/sample.txt","r") as file:
  string_list = file.readlines()

  # Total line
  total_lines = len(string_list)
  print(total_lines)

  # Total words
  number_of_words = list(map(lambda x : len(x.split()),string_list))
  total_num_of_words = reduce(lambda x,y : x+y , number_of_words)
  print(total_num_of_words)

  # Cleaning Process

  # new line remove
  string_list = list(map(str.strip,string_list))
  # space remove
  string_list = list(map(lambda x : x.replace(" ",""),string_list))
  # full_stop remove
  string_list = list(map(lambda x : x.replace(".",""),string_list))
  # Comma remove
  string_list = list(map(lambda x : x.replace(",",""),string_list))
  print(string_list)

  ## number of char calculate
  num_of_char = list(map(lambda x : len(x),string_list))
  print(num_of_char)
  total_num_of_char = reduce(lambda x,y : x+y,num_of_char)
  print(total_num_of_char)


 ## Now store every calculated thing into a file name counter_of_string.txt
with open("./sample_data/counter_of_string.txt","w") as file:
  file.write(f"Total number of lines : {total_lines}\nTotal number of words : {total_num_of_words}\nTotal number of char : {total_num_of_char}")
```

# Problem 1: Common Friends (Set Operations)
# Problem :
 Given two sets of friends from two people, find the mutual friends, unique friends of each, and the total number of unique friends.


# Example Input:

a_friends = {"Rahim", "Karim", "Sakib", "Jamal"}

b_friends = {"Sakib", "Jamal", "Rafiq", "Nadim"}

# Expected Output:

Mutual friends: {'Sakib', 'Jamal'}

Unique to A: {'Rahim', 'Karim'}

Unique to B: {'Rafiq', 'Nadim'}

Total unique friends: 4


```python
# Taking input as a list
a_friend = input("Enter element").split()
b_friend = input("Enter element").split()

# Now convert it into set
a_friend = set(a_friend)
b_friend = set(b_friend)

# Now operations
mutual_friends = a_friend & b_friend
unique_to_A = a_friend - b_friend
unique_to_B = b_friend - a_friend
total_unique_friends = a_friend | a_friend

# Now print them
print("Mutual frinds : ",mutual_friends)
print("Unique to A : ",unique_to_A)
print("Unique to B : ",unique_to_B)
print("Total unique friends : ",len(total_unique_friends))

```
