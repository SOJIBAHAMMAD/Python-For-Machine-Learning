# **Introduction to NumPy**
NumPy is a Python library used for working with arrays.
It also has functions for working in domain of linear algebra, fourier transform, and matrices.In Python we have lists that serve the purpose of arrays, but they are slow to process.The array object in NumPy is called ndarray, it provides a lot of supporting functions that make working with ***ndarray*** very easy.NumPy arrays are stored at one continuous place in memory unlike lists, so processes can access and manipulate them very efficiently.

```python
arr = [1,2,3,4,5]
print(arr*5)
arr = [x*5 for x in arr]
print(arr)
```

```python
import numpy as np
arr1 = np.array([1,2,3,4,5]) # SIMD(Single Input Multiple Data) process
print(arr1*5)
```

# Ndarray

```python
import numpy as np
# 1D array
arr1 = np.array([1,2,3,4,5])
print(type(arr1))
print(arr1.ndim) # array dimention check

# 2D array
arr2 = np.array([[1,2,3,4],[3,4,5,6]])
print(arr2.ndim) # array dimention check


# 3D array
arr3 = np.array([[[1,2,3],[2,3,4]],[[4,5,6],[5,6,7]]])
print(arr3.ndim) # array dimention check

```

# Array attributes

```python
# Dimension check
print(arr1.ndim)
print(arr2.ndim)
print(arr3.ndim)
print("\n")

# Shape
print(arr1.shape)
print(arr2.shape)
print(arr3.shape)
print("\n")

# data type
print(arr1.dtype)
print(arr2.dtype)
print(arr3.dtype)
print("\n")

# Size
print(arr1.size)
print(arr2.size)
print(arr3.size)
print("\n")
```

# ndarray data **type**

```python
arr = np.array([1,2,3,4])
print(arr.dtype)
arr = np.array([1.4,2.5,3.1,4.3])
print(arr.dtype)
```

```python
# upcasting
arr = np.array([1,2,3.5,4])
print(arr.dtype) # type cast int to float

arr = np.array([1,2,3,4.5,'a']) # typecast int to float to string
print(arr.dtype)
```

# Selecting a data type for an **array**

```python
# by-default int64
arr = np.array([1,2,3,4])
print(arr.dtype)
# way-1
arr = np.array([1,2,3,4],dtype=np.int8)
print(arr.dtype)

# way-2
arr = arr.astype(np.int16)
print(arr.dtype)
```

# ndarray creating from existing **data**

```python
# from list
lst = [1,2,3,4,5.5]
# arr = np.array(lst)
arr = np.array(lst,dtype=np.int32)
print(arr)
print(type(arr))
print(arr.dtype)
print()

mixed_lst = [10,20.4,True,'Hello']
arr1 = np.array(mixed_lst) # upcasted to string(U32 means String)
print(arr1)
print(type(arr1))
print(arr1.dtype)
print() # For space
```

```python
# from matrix
matrix = np.array([[1,2,3,4],[3,4,5,6]])
arr = np.array(matrix)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print() # For space
```

```python
# From tuple
tpl = (10,20,30,40)
arr = np.array(tpl)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print() # For space
```

```python
# From Set
st = {2,3,6,4,9}
arr = np.array(list(st),dtype=np.int16)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print() # For space
```

```python
# From Dictionary
dic = {"name" : "SAJIB AHAMMAD" , "Age" : "23" , "University" : "Northern Uni"}
keys = dic.keys()
values = dic.values()
items = dic.items()

# keys to ndarray
arr = np.array(list(keys))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print() # For space

# values to ndarray
arr = np.array(list(values))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print() # For space


# items to ndarray
arr = np.array(list(items))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print() # For space
```

# **Creating ndarray from Scratch**

```python
# np.zeros

arr = np.zeros((2,3))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
print()

arr = np.zeros_like(arr3) # same as arr3
print(arr)
print(arr.shape)
```

```python
# np.ones
arr = np.ones((4,3),dtype=np.int16)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
print()


arr = np.ones_like(arr3) # same as arr3
print(arr)
print(arr.shape)
```

```python
# np.empty
arr = np.empty((4,3),dtype=np.int8)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
print()


arr = np.empty_like(arr3) # same as arr3
print(arr)
print(arr.shape)
```

```python
# np.full ((shape), value)
arr = np.full((4,3),np.inf) # infinity value
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
print()

arr = np.full_like(arr3,10.4) # same as arr3
print(arr)
print(arr.shape)
```

# **Array Creation with random values**:

```python
# np.random.rand(shape)
arr = np.random.rand(3,4) # 0 to 1 random values
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.random.randint(stating , end(excluded) , shape)
arr = np.random.randint(1,10,(3,4)) # 1 to 10 int values
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.random.uniform(stating , end , shape)
arr = np.random.uniform(10,50,(4,3)) # 10 to 50 float values
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

# **Array creation with range function**

```python
# np.arange(start , end(excluded) , steps)

arr = np.arange(1,10,1)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
print()


# If we want a matrix
# arr = np.arange(1,10,1).reshape(3,3) or
mat = arr.reshape(3,3)
print(mat)
```

```python
# np.linspace(start , end(include) , num of points)
arr = np.linspace(1,10,15)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
print()
```

```python
# np.logspace(start , end(include) , num of points)
arr = np.logspace(0,4,5)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
print()

# we can change base
arr = np.logspace(0,4,5, base=2)
print(arr)
```

# **Creating Matrix for linear algebra**

```python
# Diagonal Matrix
# Syntax : np.diag(list)
diagonal_matrix = np.diag([2,3,4,5])
print(diagonal_matrix)
print(diagonal_matrix.shape)
```

```python
# Identity Matrix
# Syntax : np.eye(size)
identity_matrix = np.eye(4)
print(identity_matrix)
print(identity_matrix.shape)
print()

# Syntax : np.eye(row,colomn)
identity_matrix = np.eye(3,4)
print(identity_matrix)
print()


# Syntax : np.eye(row,colomn,k)
identity_matrix = np.eye(3,4,-1)
print(identity_matrix)
print()

# Syntax : np.eye(row,colomn,k)
identity_matrix = np.eye(3,4,1)
print(identity_matrix)
print()
```

# **Indexing and Slicing**

```python
# Indexing
print(arr1)
print(arr2)

print(arr1[1])
arr1[1]=20
print(arr1)


print(arr2[1][1])
arr2[1][3]=7
print(arr2)
```

```python
# Slicing
# 1D Syntax : arr[start:end:step]

print(arr1[1:4:1])
```

```python
# 2D Syntax : arr[row_start : row_end : step , colomn_start : colomn_end : step]
print(arr2)
# Getting a row
row_0 = arr2[0:1, ]
row_1 = arr2[1: , ]
print(row_0)
print(row_1)
print()


# Getting a colomn
colomn_0 = arr2[:: ,0:1]
colomn_1 = arr2[:: , 1:2]
print(colomn_0)
print(colomn_1)

# Now all together
slicing_2D = arr2[0:2 , 0:2]
print(slicing_2D)
```
