# **ndarray Manipulation**

```python
import numpy as np

arr = np.random.randint(1,100, size=(10,5))  # Sample ->10 and Feature -> 5
print(arr.shape)
print(arr)
```

```python
# If anyone ask for specific Shape
b = arr.reshape(5,10)
print(b)
print(b.shape)
print(b.ndim)
```

```python
# Converted into 1D array
flatten = b.flatten()
print(flatten.ndim)
print(flatten)
print()


# Colomn-wise flattening
Colomn_wise_flattening = np.ravel(b, order='F')
print(Colomn_wise_flattening)
print()

# 2nd way of row-wise flattening
row_wise_flattening = np.ravel(b, order='C')
print(row_wise_flattening)
```

# **Statistical Function**

```python
# data load using np.genfromtxt() function
data = np.genfromtxt('student_scores.csv' , delimiter=',', skip_header=1)
print(data)
# It's a data-set of student scores of math , science and english
```

```python
# Find Maximum & Minimum & Avg & Median & Standard Deviation math marks
math_marks = data[::, :1]
print(math_marks.T) # T means show data in a row and for showing data in colomn we use C

# Max_score
max_math_score = np.max(math_marks)
print(max_math_score)

# Min_score
min_math_score = np.min(math_marks)
print(min_math_score)

# Avg_score
avg_math_score = np.average(math_marks) # we can use both average or mean function
print(avg_math_score)

# Median_of_score
median_of_math_score = np.median(math_marks)
print(median_of_math_score)

# Standard_deviation
standard_deviation_of_math_score = np.std(math_marks)
print(standard_deviation_of_math_score)
```

# **Linear Algebra**

```python
A = np.array([[3,4,7],
              [7,3,9]])

B = np.array([[4,3,6],
              [9,1,4],
              [3,8,9]])
# Dot product -> np.dot(matirx1 , matrix2)
dot_product = np.dot(A,B)
print(dot_product)

# Trace (sum of primary diagonal value)
trace = np.trace(B)
print(trace)

# determinate -> np.linalg.det(matirx)
det_val = np.linalg.det(B)
print(det_val)

# Rank of a matrix -> np.linalg.matrix_rank(matrix)
rank = np.linalg.matrix_rank(B)
print(rank)
```

Q1: Write a Python function that calculates the mean of a matrix either by row or by column, based on a given mode. The function should take a matrix (list of lists) and a mode ('row' or 'column') as input and return a list of means according to the specified mode.

```python
import numpy as np

def calculate_matrix_mean(matrix: list[list[float]], mode: str) -> list[float]:

	mat = np.array(matrix)

	if mode == "row":
		means = np.mean(matrix,axis=1)
	else:
		means = np.mean(matrix,axis=0)

	return means
```

Q2: Write a Python function that computes the dot product of a matrix and a vector. The function should return a list representing the resulting vector if the operation is valid, or -1 if the matrix and vector dimensions are incompatible. A matrix (a list of lists) can be dotted with a vector (a list) only if the number of columns in the matrix equals the length of the vector. For example, an n x m matrix requires a vector of length m.

```python
import numpy as np

def matrix_dot_vector(a: list[list[int|float]], b: list[int|float]) -> list[int|float]:


	a = np.array(a)
	b = np.array(b)

	shape_a = a.shape
	if shape_a[1] != b.size: # shape_a[1] means colomn and size of b means row
		return -1

	dot = np.dot(a,b).tolist()

	return dot



	# Return a list where each element is the dot product of a row of 'a' with 'b'.
	# If the number of columns in 'a' does not match the length of 'b', return -1.
```

Q3 : Write a Python function that computes the transpose of a given 2D matrix. The transpose of a matrix is formed by turning its rows into columns and columns into rows. For an m×n matrix, the transpose will be an n×m matrix.

```python
import numpy as np
def transpose_matrix(a: list[list[int|float]]) -> list[list[int|float]]:

    mat = np.array(a)
    transpose = np.transpose(mat).tolist()
    return transpose

```
