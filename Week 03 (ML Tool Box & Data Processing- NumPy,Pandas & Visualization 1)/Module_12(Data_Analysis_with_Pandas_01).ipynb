# **Pandas Introduction :**   
### Pandas is a Python library used for working with data sets. It has functions for analyzing, cleaning, exploring, and manipulating data. It provides fast and flexible tools to work with tabular data, similar to spreadsheets or SQL tables.

# **What is a DataFrame?**

### A Pandas DataFrame is a 2 dimensional data structure, like a 2 dimensional array, or a table with rows and columns.

# **What is a Series?**

### A Pandas Series is like a column in a table. It is a one-dimensional array holding data of any type.

# **Transforming to DataFrame from Inbuilt datas**

```python
# From List
my_list = [['Alice',25],['Bob',23],['Charlei',28]]
list_df = pd.DataFrame(my_list,columns=['Name',"Age"],index=[1,2,3])
print(list_df)
print(type(list_df))
```

```python
# From Tuple
my_tuple = (('Alice',25),('Bob',26),('Charlie',28))
tuple_df = pd.DataFrame(my_tuple,columns=['Name','Age'],index=[1,2,3])
print(tuple_df)
print(type(tuple_df))
```

```python
# From Dictionary
my_dic = { 'Name': ['Alice','Bob','Charlie'],
          'Age': [25,26,28] }
dic_df = pd.DataFrame(my_dic)
print(dic_df)
print(type(dic_df))

# Here keys will be converted into header row
```

```python
# From dictionary of list
my_data = [ {'Name':'Alice','Age':'25','City':'NewYork'},
            {'Name':'Bob','Age':'26','City':'Paris'},
            {'Name':'Charlie','Age':'28'} ]
my_data_df = pd.DataFrame(my_data)
print(my_data_df)
```

# Deleteing rows & Columns , Changing Values and Iteration

```python
df
```

```python
# Row deletion
# Syntax : df.drop(index,inpalce = True)
# This fun return a copy but if we want real change we can use inpalce = True
row_delete = df.drop(19)
print(row_delete.to_string())
```

```python
# Column deletion
# Syntax : df.drop('Column_Name',axis = 1,inpalce = True)
# This fun return a copy but if we want real change we can use inpalce = True
column_delete = df.drop('Instructor',axis=1)
print(column_delete.to_string())
```

# **Sorting a DataFrame**

```python
# Sort Data Structure Marks Descending order
sorted_df = df.sort_values(['Data Structure Marks', 'Python Marks'], ascending = False)
sorted_df
```

```python
# Sort Data Structure Marks Descending order but Python Marks in Ascending Order
sorted_df1 = df.sort_values(['Data Structure Marks', 'Python Marks'], ascending = [0,1])
sorted_df1
```

# **Filtering Data based on Conditions**

```python
df
```

```python
# Filtering the Student who are not started yet
not_started = df.loc[df['CompletionStatus'] == 'Not Started']
not_started
```

```python
# Filtering the Student who are Completed the course
course_completed  = df.loc[df['CompletionStatus'] == 'Completed']
course_completed
```

```python
# Filtering the Student who are In Progress
In_progress  = df.loc[df['CompletionStatus'] == 'In Progress']
In_progress
```

```python
# Filters who's CompletionStatus is Completed and Data Structure Marks is >= 90

filtered_data = df.loc[ (df['CompletionStatus'] == 'Completed') & (df['Data Structure Marks'] >= 90) ]
filtered_data
```

Question : https://leetcode.com/problems/big-countries/description/?envType=study-plan-v2&envId=30-days-of-pandas&lang=pythondata

```python
import pandas as pd

def big_countries(world: pd.DataFrame) -> pd.DataFrame:
    big_country = world.loc[ (world['area'] >= 3000000) | (world['population'] >= 25000000) , ['name','population','area'] ]
    return big_country
```
