# **Filtering on String**

```python
# Filters the City that contains New word
df.loc[df["City"].str.contains("New")]  # This Syntax is case sensetive
```

```python
# Filters the City that contains New or new word (Not case sensetive)
df.loc[df["City"].str.contains("new",case=False)] # This Syntax is not case sensetive
```

```python
# Filters the City that contains Los in it's name
df.loc[df["City"].str.contains(r"Los")]
```

```python
# Filters the City that Start with Los word
df.loc[df["City"].str.contains(r"^Los")]
```

```python
# Filters the City that end with rk
df.loc[df["City"].str.contains(r"rk$")]
```

```python
# Filters the City that contains New and Los
df.loc[df["City"].str.contains(r"New|Los")]
```

# **Apply function on DataFrame**

```python
df
```

```python
# min & max scaling

mx = df['Total Marks'].max()
mn = df['Total Marks'].min()

df['Scaled Marks'] = df['Total Marks'].apply(lambda x : (x-mn)/(mx-mn))
df
```

```python
# Custom Function

def grading_system(marks):
  if marks >= 260:
    return "A+"
  elif marks >= 250:
    return "A"
  else:
    return 'A-'


df['Grade'] = df['Total Marks'].apply(grading_system)
df
```

```python
# we can apply it into entire DataFrame

def marking_system(df):
  a = df['Data Structure Marks'] * 2
  b = df['Algorithm Marks'] * 3
  c = df['Python Marks'] * 4

  return a+b+c

df['Exceptional Marks'] = df.apply(marking_system,axis=1)
df
```

```python
import pandas as pd

def delete_duplicate_emails(person: pd.DataFrame) -> None:

    person.sort_values('id',inplace=True)
    person.drop_duplicates(subset='email',inplace=True)
```

## Problem-2 : https://leetcode.com/problems/count-salary-categories/description/?envType=study-plan-v2&envId=30-days-of-pandas&lang=pythondata

```python
import pandas as pd

def count_salary_categories(accounts: pd.DataFrame) -> pd.DataFrame:

    low_salary = (accounts['income'] < 20000).sum()
    avg_salary = ((accounts['income'] >= 20000) & (accounts['income'] <= 50000)).sum()
    high_salary = (accounts['income'] > 50000).sum()

    data = [('Low Salary',low_salary),('Average Salary',avg_salary),('High Salary',high_salary)]

    df = pd.DataFrame(data,columns=['category','accounts_count'])
    return df
```
