# **2D line plot**

```python
hours = [2,5,3,1,3,7,1]
days = [1,2,3,4,5,6,7]
plt.xlabel('Days')
plt.ylabel('Study Hours')
plt.title('Study trend over days')

# plt.plot(x,y) function
plt.plot(days,hours)
```

# **2D line plot from Pandas DataFrame**

```python
df = pd.read_csv('enrollment_data.csv')
df
```

```python
plt.xlabel('Years')
plt.ylabel('Enrollment students')
plt.title('Student enrollment over years')

# plt.plot(x,y) function
plt.plot(df['Year'],df['Programming'])
plt.plot(df['Year'],df['Digital Marketing'])
```

```python
# Adding label using plt.legend() function

plt.xlabel('Years')
plt.ylabel('Enrollment students')
plt.title('Student enrollment over years')

# plt.plot(x,y,label) function
plt.plot(df['Year'],df['Programming'], label='Programming')
plt.plot(df['Year'],df['Digital Marketing'], label='Digital Marketing')
plt.legend()
```

```python
# More visualizations

plt.xlabel('Years')
plt.ylabel('Enrollment students')
plt.title('Student enrollment over years')

# plt.plot(x,y,label,color,linewidth,linestyle,marker,markersize) function
plt.plot(df['Year'],df['Programming'], label='Programming',color='red',linewidth=2,marker='o')
plt.plot(df['Year'],df['Digital Marketing'], label='Digital Marketing',color='green',linewidth=2,marker='o')
plt.legend()
plt.grid()
```

# **Scatter Plot** :
### Scatter plots are one of the most fundamental tools for visualizing relationships between two numerical variables. matplotlib.pyplot.scatter() plots points on a Cartesian plane defined by X and Y coordinates. Each point represents a data observation, allowing us to visually analyze how two variables correlate, cluster or distribute.

```python
student = pd.read_csv('student_IQdata.csv')
student
```

```python
# Relation between Study Hour and IQ Score -> Positive Correlation
plt.xlabel('Study hour')
plt.ylabel('IQ Score')
plt.title('IQ Score over Study Hour')

# plt.scatter(x,y) function
plt.scatter(student['Study_Hour'],student['IQ_Score'])
plt.show()
```

```python
# Relation between Shoe_Size and IQ Score -> No Correlation
plt.xlabel('Shoe Size')
plt.ylabel('IQ Score')
plt.title('IQ Score over Shoe Size')

# plt.scatter(x,y) function
plt.scatter(student['Shoe_Size'],student['IQ_Score'])
plt.show()
```

```python
# Relation between Chilling_Hours and IQ Score -> Negative Correlation
plt.xlabel('Chilling Hour')
plt.ylabel('IQ Score')
plt.title('IQ Score over Chilling Hour')

# plt.scatter(x,y) function
plt.scatter(student['Chilling_Hours'],student['IQ_Score'])
plt.show()
```

#**Histogram** :

# **Bar Chart**

```python
data = {
    'Student': ['B', 'C', 'D', 'E', 'F', 'G', 'H'],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male'],
    'StudyHours': [2, 5, 3, 7, 4, 8, 1],
    'Attendance': [60, 88, 70, 95, 80, 98, 55],
    'Grade': ['Fail', 'Pass', 'Fail', 'Pass', 'Pass', 'Pass', 'Fail']
}

df = pd.DataFrame(data)
df
```

```python
# Based on Gender

gender_group = df.groupby('Gender').size()
print(gender_group)
print()

print(gender_group.index)
print()
print(gender_group.values)
```

```python
plt.xlabel('Gender')
plt.ylabel('Count')
plt.title('Bar chart of Gender among Student')

# plt.bar(x,y) function
plt.bar(gender_group.index, gender_group.values)
plt.show()
```

```python
# Bar chart on real data

df = pd.read_csv('student_data.csv')
df
```

```python
# Based on CompletionStatus
CompletionStatus_group = df.groupby('CompletionStatus').size()
print(CompletionStatus_group)
print()
print(CompletionStatus_group.index)
print()
print(CompletionStatus_group.values)
```

```python
plt.xlabel('Course Completion Status')
plt.ylabel('Count')
plt.title('Bar chart of Course Completion Status among Student')

# plt.bar(x,y,color,edgecolor) function
plt.bar(CompletionStatus_group.index, CompletionStatus_group.values, color = ['green','orange','red'], edgecolor='black')
plt.show()
```

```python
# Based on Location

location_group = df.groupby('Location').size()
print(location_group)
print()

print(location_group.index)
print()

print(location_group.values)
print()
```

```python
plt.xlabel('Locations')
plt.ylabel('Count')
plt.title('Bar chart of Locations among Student')

# plt.bar(x,y,color,edgecolor) function
plt.bar(location_group.index, location_group.values,color = ['green','orange','red','yellow','blue'], edgecolor='black')
plt.show()
```

# **Pie Chart**

### A pie chart is a circular graph used to represent data as slices of a circle, where each slice corresponds to a category's proportion of the whole. It is commonly used to compare the percentage or share of different categories in a dataset.

```python
df = pd.read_csv('student_data.csv')

# Based on CompletionStatus
CompletionStatus_group = df.groupby('CompletionStatus').size()
print(CompletionStatus_group)
```

```python
plt.pie(CompletionStatus_group , labels=CompletionStatus_group.index, autopct='%1.1f%%')
plt.show()
```
