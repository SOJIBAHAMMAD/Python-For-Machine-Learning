```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df = pd.read_csv('Assignment_dataset.csv')
print(df.to_string())
```

# **Answer to the Q1 :**

```python
arr = np.array(df['TrainingHours'], dtype=np.float64)
mean = np.mean(arr)
std = np.std(arr)
print(mean)
print(std)
```

# **Answer to the Q2 :**

```python
df['Salary'] = df['Salary'].fillna(df.groupby('Department')['Salary'].transform('median'))
```

```python
total_salary = df.loc[(df['ExperienceYears'] > 12) & (df['ProjectCount']>5),'Salary'].sum()
print(total_salary)
```

# **Answer to the Q3 :**

```python
ans = df[ (df['Department'].isin(['IT','Finance'])) & (df['Age'].between(30,45)) &
          (df['PerformanceScore'] > 88)].sort_values('Salary',ascending=False)

print(ans[['Name','Department','Salary','PerformanceScore']])
```

# **Answer to the Q4 :**

```python
df['SalaryPerHour'] = (df['Salary']) / (df['WorkHoursPerWeek'] * 4)
ans = df.sort_values('SalaryPerHour',ascending=False)
print(ans[['Name','SalaryPerHour']].head())
```
